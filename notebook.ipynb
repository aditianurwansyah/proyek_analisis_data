## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.

- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

Karena sifat datanya yang berkelanjutan (time-series per jam), pengisian data kosong (imputasi) menggunakan metode Interpolasi (Linear) lebih direkomendasikan dibandingkan menggunakan nilai rata-rata, agar alur waktu polusi tetap natural.

- Inconsistent value: Pada Kolom penunjuk waktu masih terpecah ke dalam 4 kolom terpisah, yaitu year, month, day, dan hour, yang semuanya masih bertipe data angka integer (int64), belum digabung menjadi format waktu resmi (datetime).

### Cleaning Data

#### Menggabungkan Kolom Waktu dan Menangani Missing Value

```python
# Menggabungkan kolom waktu menjadi satu kolom datetime
df['datetime'] = pd.to_datetime(df[['year', 'month', 'day', 'hour']])
```

```python
# Cek informasi tipe data terbaru
df.info()
```

```python
# Daftar kolom polutan dan cuaca yang memiliki missing values
columns_to_interpolate = ['PM2.5', 'PM10', 'SO2', 'NO2', 'CO', 'O3', 'TEMP', 'PRES', 'DEWP', 'RAIN', 'WSPM']

# Melakukan interpolasi linear untuk kolom numerik
df[columns_to_interpolate] = df[columns_to_interpolate].interpolate(method='linear')

# Untuk kolom kategori teks (seperti arah angin 'wd'), kita bisa mengisi dengan nilai yang sering muncul (modus)
df['wd'] = df['wd'].fillna(df['wd'].mode()[0])

# Cek apakah masih ada missing values
print("Sisa Missing Values:\n", df.isna().sum())
```

**Insight:** (Opsional)
- Pada kolom year, month, day dan hour itu sudah digabungkan menjadi tipe data yang baru yaitu datetime. Sehingga dengan perubahan ini dapat memudahkan analisis tren berbasis waktu.

- Seluruh missing values pada parameter polutan dan cuaca telah berhasil diatasi menggunakan metode Interpolasi Linear dan pengisian modus untuk data kategori, sehingga kontinuitas data per jam tetap terjaga dengan baik tanpa merusak pola tren aslinya.

## Visualization & Explanatory Analysis

### Pertanyaan 1: Bagaimana tren rata-rata tingkat konsentrasi PM2.5 bulanan di stasiun Aotizhongxin sepanjang tahun 2014 hingga 2016, dan pada bulan apa polusi mencapai titik tertinggi?

```python
df_14_16 = df[(df['year'] >= 2014) & (df['year'] <= 2016)]
tren_bulanan = df_14_16.groupby('month')['PM2.5'].mean().reset_index()

# Menampilkan hasil output EDA berupa tabel
display(tren_bulanan)
```

```python
# Visualization
plt.figure(figsize=(10, 5))
sns.lineplot(data=tren_bulanan, x='month', y='PM2.5', marker='o', color='red')
plt.title('Tren Rata-Rata PM2.5 Bulanan (2014-2016) di Aotizhongxin')
plt.xlabel('Bulan')
plt.ylabel('Rata-rata Konsentrasi PM2.5')
plt.xticks(range(1, 13))
plt.grid(True)
plt.show()
```

### Pertanyaan 2: Bagaimana korelasi antara kecepatan angin (WSPM) dan curah hujan (RAIN) terhadap tingkat konsentrasi PM10 di stasiun Aotizhongxin selama musim dingin (Desember–Februari) pada periode 2013–2016?

```python
##Membuat fungsi kategori kecepatan angin
musim_dingin = df[
    (df['year'] >= 2013) &
    (df['year'] <= 2016) &
    (df['month'].isin([12, 1, 2]))
]
korelasi = musim_dingin[['WSPM', 'RAIN', 'PM10']].corr()
display(korelasi)
```

```python
# Visualization (Heatmap)
plt.figure(figsize=(6, 4))
sns.heatmap(korelasi, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Korelasi Kecepatan Angin, Curah Hujan, dan PM10 (Musim Dingin)')
plt.show()
```

### Pertanyaan 3: Bagaimana perbedaan rata-rata konsentrasi PM2.5 pada jam sibuk lalu lintas (07:00–09:00 dan 17:00–19:00) dibandingkan dengan jam non-sibuk di stasiun Aotizhongxin sepanjang tahun 2015?

```python
# 1. Filter data khusus tahun 2015
df_2015 = df[df['year'] == 2015].copy()

# 2. Membuat fungsi untuk mengategorikan jam sibuk (07:00-09:00 dan 17:00-19:00)
def kategori_jam(hour):
    if (7 <= hour <= 9) or (17 <= hour <= 19):
        return 'Jam Sibuk (07:00-09:00 & 17:00-19:00)'
    else:
        return 'Bukan Jam Sibuk'

# 3. Menerapkan fungsi tersebut ke kolom baru bernama 'kategori_waktu'
df_2015['kategori_waktu'] = df_2015['hour'].apply(kategori_jam)

# 4. Menghitung rata-rata PM2.5 berdasarkan kategori waktu tersebut
rata_rata_pm25_2015 = df_2015.groupby('kategori_waktu')['PM2.5'].mean().reset_index()

# Menampilkan hasil tabelnya
display(rata_rata_pm25_2015)
```

```python
plt.figure(figsize=(8, 5))
# Membuat Bar Chart menggunakan seaborn dengan perbaikan parameter hue
sns.barplot(x='kategori_waktu', y='PM2.5', data=rata_rata_pm25_2015, hue='kategori_waktu', palette='Set2', legend=False)

plt.title('Perbandingan Rata-rata PM2.5 pada Jam Sibuk vs Non-Sibuk di Aotizhongxin (2015)')
plt.xlabel('Kategori Waktu')
plt.ylabel('Rata-rata Konsentrasi PM2.5')
plt.show()
```

**Insight:** (Opsional)
- **Pertanyaan 1 (Tren Bulanan):** Lonjakan polusi tertinggi selalu terjadi pada bulan-bulan musim dingin (Oktober hingga Februari), yang dipicu oleh tingginya aktivitas pembakaran batu bara untuk pemanas ruangan di Beijing.
- **Pertanyaan 2 (Korelasi Cuaca):** Kecepatan angin (WSPM) memiliki korelasi negatif yang cukup signifikan terhadap PM10, menunjukkan bahwa angin berperan penting dalam mendispersi polutan, sementara curah hujan pada musim dingin memiliki pengaruh yang sangat kecil.
- **Pertanyaan 3 (Jam Sibuk vs Non-Sibuk):** Pada tahun 2015, rata-rata PM2.5 pada jam sibuk (77.26 µg/m³) secara mengejutkan sedikit lebih rendah dibandingkan jam non-sibuk (83.36 µg/m³). Hal ini mengindikasikan bahwa akumulasi polusi di Aotizhongxin tidak hanya dipengaruhi oleh volume lalu lintas lokal, tetapi juga oleh faktor dispersi atmosfer dan polusi kiriman.

## Analisis Lanjutan (Opsional)

```python
# Mengelompokkan Kualitas Udara berdasarkan nilai PM2.5 (Binning)
def categorize_pm25(pm):
    if pm <= 35:
        return 'Baik'
    elif pm <= 75:
        return 'Sedang'
    elif pm <= 150:
        return 'Tidak Sehat'
    else:
        return 'Berbahaya'

df['kategori_udara'] = df['PM2.5'].apply(categorize_pm25)
df['kategori_udara'].value_counts()
```

```python
df['kategori_udara'] = df['PM2.5'].apply(categorize_pm25)
jumlah_kategori = df['kategori_udara'].value_counts()
urutan = ['Baik', 'Sedang', 'Tidak Sehat', 'Berbahaya']
jumlah_kategori = jumlah_kategori.reindex(urutan, fill_value=0)
warna = ['#28a745', '#ffc107', '#fd7e14', '#dc3545']
plt.figure(figsize=(10, 6))

# Membuat Bar Chart
ax = sns.barplot(
    x=jumlah_kategori.index,
    y=jumlah_kategori.values,
    hue = jumlah_kategori.index,
    palette=warna,
    legend=False
)

# Menambahkan Judul dan Label Sumbu agar informatif
plt.title('Kategori Kualitas Udara (Berdasarkan PM2.5)', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Kategori Kualitas Udara', fontsize=12, labelpad=10)
plt.ylabel('Jumlah Observasi', fontsize=12, labelpad=10)

# Menambahkan teks angka persis di atas setiap batang grafik
for p in ax.patches:
    # Memastikan angka tidak ditampilkan jika nilainya 0 (opsional untuk kerapian)
    if p.get_height() > 0:
        ax.annotate(f'{int(p.get_height())}',
                    (p.get_x() + p.get_width() / 2., p.get_height()),
                    ha='center', va='bottom',
                    fontsize=11, fontweight='bold', color='black',
                    xytext=(0, 5), textcoords='offset points')

# Menambahkan garis bantu (grid) horizontal tipis di latar belakang
plt.grid(axis='y', linestyle='--', alpha=0.6)
sns.despine() # Menghapus garis tepi atas dan kanan bingkai

# Menampilkan hasil visualisasi
plt.tight_layout()
plt.show()
```

**Insight:** (Opsional)
- Sepanjang tahun observasi, rekam kualitas udara paling sering berada pada kategori 'Baik', yaitu sebanyak 12.447 observasi (sekitar 35,5% dari total 35.064 observasi). Angka ini menjadi indikator positif bahwa masih terdapat banyak waktu di mana udara berada pada titik aman untuk aktivitas warga di luar ruangan.
